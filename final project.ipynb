```python
import pandas as pd
import numpy as np

# 1. Load original dataset
df = pd.read_csv("diabetes_dataset.csv")  # change path if needed

# 2. Columns that are invasive (blood-test based)
invasive_cols = [
    "cholesterol_total",
    "hdl_cholesterol",
    "ldl_cholesterol",
    "triglycerides",
    "glucose_fasting",
    "glucose_postprandial",
    "insulin_level",
    "hba1c",
]

# 3. Drop invasive features
df_noninv = df.drop(columns=invasive_cols)

# 4. Define target columns that we DON'T want to accidentally drop
target_cols = ["diagnosed_diabetes", "diabetes_stage", "diabetes_risk_score"]

# 5. Select numeric feature columns (excluding targets)
numeric_cols = df_noninv.select_dtypes(include=[np.number]).columns
feature_cols = [c for c in numeric_cols if c not in target_cols]

# 6. Correlation matrix on numeric predictors
corr = df_noninv[feature_cols].corr().abs()

# 7. Keep only upper triangle of correlation matrix
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))

# 8. Choose a threshold and collect features to drop
threshold = 0.85   # you can change this to 0.8 / 0.9 etc.
to_drop_corr = [col for col in upper.columns if any(upper[col] > threshold)]

print("Highly correlated features to drop:", to_drop_corr)

# 9. Drop the highly correlated features
df_reduced = df_noninv.drop(columns=to_drop_corr)

# 10. Save outputs
df_noninv.to_csv("diabetes_noninvasive.csv", index=False)           # only non-invasive
df_reduced.to_csv("diabetes_noninvasive_reduced.csv", index=False)  # + correlation filter

print("Original shape:         ", df.shape)
print("Non-invasive shape:     ", df_noninv.shape)
print("Reduced (no high corr): ", df_reduced.shape)

```

```python
import pandas as pd

# 1. Load raw dataset
df = pd.read_csv("diabetes_dataset.csv")  # adjust path if needed

# 2. Invasive (blood-test) features to remove
invasive_cols = [
    "cholesterol_total",
    "hdl_cholesterol",
    "ldl_cholesterol",
    "triglycerides",
    "glucose_fasting",
    "glucose_postprandial",
    "insulin_level",
    "hba1c",
]

# 3. Drop invasive features + extra targets you don't want
drop_cols = invasive_cols + ["diabetes_stage", "diabetes_risk_score"]
df = df.drop(columns=drop_cols)

# 4. Separate target
target_col = "diagnosed_diabetes"
y = df[target_col]
X = df.drop(columns=[target_col])

# 5. One-hot encode categorical predictors
# (object or category dtypes)
X_encoded = pd.get_dummies(X, drop_first=True)

# 6. Reattach target
df_classification = X_encoded.copy()
df_classification[target_col] = y

# 7. Save to new file
output_path = "diabetes_classification_noninvasive.csv"
df_classification.to_csv(output_path, index=False)

print("Saved clean classification dataset to:", output_path)
print("Shape:", df_classification.shape)
print("Columns:\n", df_classification.columns.tolist())
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

# 1. Load the prepared classification dataset
df = pd.read_csv("diabetes_classification_noninvasive.csv")
print(df.head())
target = "diagnosed_diabetes"
X = df.drop(columns=[target])
y = df[target]

# 2. Train–test split
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.2, random_state=0, stratify=y)

# 3. Logistic Regression model (recommended baseline)
logit_clf = Pipeline([
    ("scaler", StandardScaler()),  # scales all numeric/dummy features
    ("model", LogisticRegression(max_iter=1000, class_weight="balanced"))
])

logit_clf.fit(X_train, y_train)
y_pred_logit = logit_clf.predict(X_test)

print("=== Logistic Regression ===")
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred_logit))
print("\nClassification report:\n", classification_report(y_test, y_pred_logit))

# 4. Random Forest model (non-linear, often stronger)
rf_clf = RandomForestClassifier(
    n_estimators=300,
    random_state=0,
    class_weight="balanced",
    n_jobs=-1
)

rf_clf.fit(X_train, y_train)
y_pred_rf = rf_clf.predict(X_test)

print("=== Random Forest ===")
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred_rf))
print("\nClassification report:\n", classification_report(y_test, y_pred_rf))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier

# 1. Load your prepared classification dataset
df = pd.read_csv("diabetes_classification_noninvasive.csv")

target = "diagnosed_diabetes"
X = df.drop(columns=[target])
y = df[target]

# 2. Train–test split (stratified to keep class ratio)
X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.2,random_state=0, stratify=y)

# ==============================
# 3. Decision Tree classifier
# ==============================
dt_clf = DecisionTreeClassifier(
    max_depth=6,          # limit depth to reduce overfitting
    min_samples_split=50, # require some samples to split
    random_state=0,
    class_weight="balanced"   # handle class imbalance a bit
)

dt_clf.fit(X_train, y_train)
y_pred_dt = dt_clf.predict(X_test)

print("=== Decision Tree ===")
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred_dt))
print("\nClassification report:\n", classification_report(y_test, y_pred_dt))

# ==============================
# 4. AdaBoost classifier
# ==============================
# Uses decision stumps (max_depth=1 trees) internally by default
ada_clf = AdaBoostClassifier(
    n_estimators=200,   # number of weak learners
    learning_rate=0.1,
    random_state=0
)

ada_clf.fit(X_train, y_train)
y_pred_ada = ada_clf.predict(X_test)

print("=== AdaBoost ===")
print("Confusion matrix:\n", confusion_matrix(y_test, y_pred_ada))
print("\nClassification report:\n", classification_report(y_test, y_pred_ada))
```

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import make_scorer, recall_score, f1_score

# 1. Load your prepared dataset
df = pd.read_csv("diabetes_classification_noninvasive.csv")

target = "diagnosed_diabetes"
X = df.drop(columns=[target])
y = df[target]

# 2. Define models (you can add others)
models = {
    "logistic": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=1000, class_weight="balanced"))
    ]),
    "random_forest": RandomForestClassifier(
        n_estimators=300, random_state=0, class_weight="balanced", n_jobs=-1
    ),
    "decision_tree": DecisionTreeClassifier(
        max_depth=6, min_samples_split=50, class_weight="balanced", random_state=0
    ),
    "adaboost": AdaBoostClassifier(
        n_estimators=200, learning_rate=0.1, random_state=0
    ),
}

# 3. CV setup and scoring
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)

scoring = {
    "accuracy": "accuracy",
    "recall_1": make_scorer(recall_score, pos_label=1),
    "f1_1": make_scorer(f1_score, pos_label=1),
}

# 4. Run CV for each model
for name, clf in models.items():
    print(f"\n=== {name} ===")
    scores = cross_validate(clf, X, y, cv=cv, scoring=scoring, n_jobs=-1)

    for metric in scoring.keys():
        vals = scores[f"test_{metric}"]
        print(f"{metric}: {vals.mean():.3f} ± {vals.std():.3f}")
```
