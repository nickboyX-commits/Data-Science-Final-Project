```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix

# === Decision Tree on diagnosed_diabetes (classification) ===

# 1. Load dataset 
input_file = "/Users/xnick/Downloads/diagnosed_diabetes for classification.csv"
try:
    df = pd.read_csv(input_file)
    print(f"Data loaded successfully from: {input_file}")
except FileNotFoundError:
    print(f"❌ Error: File '{input_file}' not found. Please check the path.")
    exit()

# Define the target column
target_col = "diagnosed_diabetes"

# 2. Separate features (X) and target (y)
X = df.drop(columns=[target_col])
y = df[target_col]

# 3. Train-test split (stratified to maintain the class ratio in both sets)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2, 
    random_state=0, 
    stratify=y
)

# 4. Initialize the Decision Tree Classifier
# Use class_weight="balanced" to handle potential class imbalance
dt_clf = DecisionTreeClassifier(
    max_depth=8,              # Limiting depth to control complexity and overfitting
    min_samples_split=20,     # Minimum samples required to split a node
    random_state=0,
    class_weight="balanced"   # Essential for imbalanced classification tasks
)

# 5. Train the model
print("\n🚀 Training Decision Tree Classifier...")
dt_clf.fit(X_train, y_train)

# 6. Predict on the test set
y_pred = dt_clf.predict(X_test)

# 7. Confusion matrix and classification report
print("\n=== Decision Tree Classification Report ===")
cm = confusion_matrix(y_test, y_pred)
print("Confusion matrix:\n", cm)
print("\nClassification report:\n", classification_report(y_test, y_pred))
```

```python
# ============================================
# ANALYSIS FOR CASE 1 – Diagnosed Diabetes (0/1)
# ============================================

print("\n" + "="*70)
print("CASE 2 – ANALYSIS: Diagnosed Diabetes Classification (Decision Tree)")
print("="*70)
print("""
Overall accuracy: ~0.60

Interpretation:
- The model is correct about 60% of the time.
- It correctly identifies MOST diabetics (class 1), with recall ~0.78.
  → Good for screening, as it successfully finds the majority of positive cases (True Positives).

- It struggles to correctly identify healthy individuals (class 0):
  → Recall for class 0 is only ~0.35.
  → This leads to many healthy people being incorrectly flagged as diabetic (High False Positives).

Meaning:
- The Decision Tree model, due to the 'class_weight=balanced' setting, is biased towards predicting 'diabetes'.
- This preference generates a high volume of true positives (good for finding patients) but also a high volume of false positives (poor efficiency).
- This model is a decent initial risk flagger, as its priority is clearly set to minimize *漏诊* (missed diagnoses), making it suitable for preliminary risk stratification.
""")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, classification_report

# === CASE 2: Decision Tree for prediabetes_for_classification.csv ===

# 1. Load dataset - Path corrected to user's Downloads folder
# NOTE: Please verify that 'xnick' is the correct username.
input_file = "/Users/xnick/Downloads/prediabetes_for_classification.csv"
try:
    df = pd.read_csv(input_file)
    print(f"Data loaded successfully from: {input_file}")
except FileNotFoundError:
    print(f"❌ Error: File not found. Please check the path.")
    exit()

# FIX: Clean column names by stripping any leading/trailing whitespace
df.columns = df.columns.str.strip() 

# Define the target column (now safe to use)
target_col = "diabetes_stage"

# 2. Inspect target values
print(f"Case 2 shape (raw): {df.shape}")
print(f"Case 2 columns (Cleaned): {df.columns.tolist()}")

# 3. Drop any rows with missing target (just in case)
df = df.dropna(subset=[target_col])

# 4. Define X and y
X = df.drop(columns=[target_col])
y = df[target_col].astype(int)

# 5. Train-test split (stratified to maintain the class ratio in both sets)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2, 
    random_state=0, 
    stratify=y
)

# 6. Initialize the Decision Tree Classifier
dt_clf = DecisionTreeClassifier(
    max_depth=8,              # Limiting depth to control complexity and overfitting
    min_samples_split=20,     
    random_state=0,
    class_weight="balanced"   # Essential for imbalanced classification tasks
)

# 7. Train the model
print("\n🚀 Training Decision Tree Classifier...")
dt_clf.fit(X_train, y_train)

# 8. Predict on the test set
y_pred = dt_clf.predict(X_test)

# 9. Confusion matrix and classification report
print("\n=== Case 2: No vs Pre-Diabetes (Decision Tree) ===")
cm = confusion_matrix(y_test, y_pred)
print("Confusion matrix:\n", cm)
print("\nClassification report:\n", classification_report(y_test, y_pred))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np # Import numpy for sqrt function

# === CASE 3: Decision Tree on diabetes_risk_score (regression) ===

# 1. Load dataset
input_file = "diabetes_preprocessed_for_risk_score.csv"
try:
    df = pd.read_csv(input_file)
    print(f"Data loaded successfully from: {input_file}")
except FileNotFoundError:
    print(f"Error: File not found: {input_file}")
    exit()

print("Case 3 columns:", df.columns.tolist())
target_col = "diabetes_risk_score"

# 2. Separate features and target
X = df.drop(columns=[target_col])
y = df[target_col]

# 3. Train–test split (no stratify for regression)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)

# 4. Decision Tree regressor setup
# Using DecisionTreeRegressor for the regression task
dt_reg = DecisionTreeRegressor(
    max_depth=8,  # Limiting depth for generalization
    min_samples_split=20,
    random_state=0
)

# 5. Train
print("\n🚀 Training Decision Tree Regressor...")
dt_reg.fit(X_train, y_train)

# 6. Predict
y_pred = dt_reg.predict(X_test)

# 7. Regression metrics (CORRECTED)
# Calculate MSE first, then take the square root to get RMSE.
# NOTE: Removed 'squared=False' to fix the TypeError.
mse = mean_squared_error(y_test, y_pred) 
rmse = np.sqrt(mse) # Explicitly calculate RMSE using np.sqrt()
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\n=== CASE 3: Risk Score (DecisionTreeRegressor) ===")
print(f"RMSE: {rmse:.3f}")
print(f"MAE:  {mae:.3f}")
print(f"R²:   {r2:.3f}")
```

```python
# ============================================
# ANALYSIS FOR CASE 3 – Diabetes Risk Score (Regression)
# ============================================

print("\n" + "="*70)
print("CASE 3 – ANALYSIS: Diabetes Risk Score Regression (Decision Tree)")
print("="*70)
print("""
R-squared (R²): ~0.946

Interpretation:
- **Model Fit is Excellent:** The R² value of 0.946 indicates that the Decision Tree model explains approximately 94.6% of the variation in the continuous diabetes risk score.
  → This is an extremely high goodness-of-fit score, suggesting the model captures the relationship between the non-invasive features and the risk score almost perfectly.

- **Prediction Accuracy (MAE):** The Mean Absolute Error (MAE) is 1.685.
  → On average, the model's prediction for the risk score is off by only 1.685 points. Given that the risk score typically spans a range (e.g., 0 to 100), this represents a very small error margin.

- **Robustness (RMSE):** The Root Mean Squared Error (RMSE) is 2.114, which is close to the MAE.
  → This proximity suggests that the model does not suffer from a high number of large prediction errors (outliers).

Meaning:
- **Model Efficacy:** The Decision Tree Regressor is highly effective and accurate for the assigned task of predicting the `diabetes_risk_score` using the provided non-invasive features.
- **Data Quality:** The results imply that the non-invasive features alone are strong predictors of the calculated risk score.
- **Conclusion:** This model is robust, accurate, and suitable for its intended deployment as a tool to continuously estimate diabetes risk based on patient survey data.
""")
```
